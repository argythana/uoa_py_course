The three notebooks must share a **similar, family-style name**:

All three notebooks live in **one folder**. Inside that folder there must be a separate `data/` subfolder containing the dataset files needed to run the notebooks.

## 3. Required Content Sections — Inside Each Notebook

> The 1)–7) numbering below refers to the sections **inside each of your notebooks** — not sections of *this* instructions file.

Each of the three notebooks must contain the following sections, clearly delimited with descriptive **markdown** headers. You make change the order, slighty but ensure the sections headers are accurate and appropriate.

### 1) Imports
Import the libraries needed for the notebook.

### 2) Load the data and describe the variables
Read the data file, then briefly describe:

- the **kind** of data and what it represents,
- each **feature** — whether it is *quantitative* or *categorical*, what it measures/records, and the range of values it takes,
- the **target variable** (`y`).

### 3) Exploratory Data Analysis (EDA)
Explore and visualise the features with charts that fit each variable's type (histograms, boxplots, bar charts, scatter plots, etc.). Comments may be **brief or extended**, depending on whether they help understanding.

### 4) Descriptive statistics
Summary statistics of the features, and — where it is informative — pairwise correlations.

### 5) Data preparation, preprocessing, and train/test split
- Convert features to the appropriate types if needed.
- Split the data into a **training subset** and a **test subset**.
- If appropriate, **rescale** features and/or build **derived/engineered features**.

### 6) Apply the algorithm(s)
Train the algorithm and predict on the test set. For **each** algorithm you must run **at least two different variants** (different hyperparameters, different feature sets, …). Examples:

- *Regression* — drop features that don't seem to influence the result significantly.
- *Clustering* — try more or fewer clusters.
- *KNN* — try a smaller number of neighbouring samples, etc.

The variant you ultimately consider best must be **briefly justified** in the next section ("Evaluation").

> For the **Classification** notebook this section is split into **sub-sections, one per classification algorithm**, in the order they were presented in the lectures.

After the test-set predictions, also predict the target value for **one new hypothetical observation** — i.e. construct a single new row of feature values for the matrix `X` and report what `y` the algorithm predicts for it.

### 7) Evaluation of model performance
Evaluate the model's **predictive ability** (not its speed) using the metric appropriate for the data and the algorithm. Comments should be **concise**.

For the **Classification** notebook, the evaluation must compare:

1. the variants of the **same** classification algorithm, and
2. **all** classification algorithms against each other.

After identifying the most accurate classification algorithm, add a few short comments explaining **why** you think this algorithm produces more accurate results than the others.
