### Defining Key Metrics:
- **Accuracy**: Proportion of correct recommendations out of all the recommendations made.
- **Recall**: Proportion of relevant customers that the model is able to identify out of all the relevant customers.

The relationship between accuracy and recall in terms of overall effectiveness depends on how many correct recommendations you're making compared to the size of the relevant population.

### Basic Equations
Let’s define:
- \( R \) = Total number of relevant customers in the dataset.
- \( P \) = Total number of customers you want to target with recommendations.
- **Accuracy** = \( \frac{\text{True Positives (TP)}}{\text{Total Recommendations (TP + FP)}} \)
- **Recall** = \( \frac{\text{True Positives (TP)}}{\text{Total Relevant Customers (R)}} \)

So, using these formulas:
- **True Positives (TP)**: Number of correctly recommended customers = \( \text{Recall} \times R \)
- **False Positives (FP)**: Number of incorrect recommendations = \( \frac{\text{True Positives (TP)}}{\text{Accuracy}} - \text{True Positives (TP)} \)

### Making the Models Indifferent
To make the models indifferent in terms of results, the overall **correct recommendations** made by each model (True Positives) should be the same. Let's see how this could happen mathematically.

#### For Model 1:
- Let \( Acc_1 \) be the accuracy of Model 1 and \( Rec_1 \) its recall.
- True Positives (correct recommendations) for Model 1 = \( TP_1 = Rec_1 \times R_1 \times Acc_1 \)

#### For Model 2:
- Let \( Acc_2 \) be the accuracy of Model 2 and \( Rec_2 \) its recall.
- True Positives for Model 2 = \( TP_2 = Rec_2 \times R_2 \times Acc_2 \)

For the results to be indifferent, the number of **true positives** from both models must be equal, so:

\[
Rec_1 \times R_1 \times Acc_1 = Rec_2 \times R_2 \times Acc_2
\]

In simpler terms, you are looking for the point at which the trade-off between recall and accuracy is balanced. 

### Numerical Example with Simplified Ratios
- Suppose you have **10,000 relevant customers** in Group 1 and **200,000 relevant customers** in Group 2.
- If **Model 1** has an accuracy of 90% and a recall of 50%, then the true positives are:
  \[
  TP_1 = 0.5 \times 10,000 \times 0.9 = 4,500 \, \text{correct recommendations}.
  \]
- For **Model 2**, to make the result indifferent, the number of correct recommendations should also be 4,500. If **Model 2** has a recall of 90%, then:
  \[
  4,500 = 0.9 \times 200,000 \times Acc_2
  \]
  Solving for \( Acc_2 \), we get:
  \[
  Acc_2 = \frac{4,500}{0.9 \times 200,000} = 0.025 = 2.5\%
  \]

So, if **Model 2** has an accuracy of 2.5%, it would deliver the same number of true positives as Model 1. If the accuracy is higher, then Model 2 will perform better in terms of correct recommendations, and if it's lower, Model 1 will outperform it.

### Conclusion:
The key balance point is when the product of accuracy and recall leads to the same number of true positives for both models. The larger the customer group, the more you can tolerate lower accuracy if the recall is high enough. Conversely, in smaller groups, higher accuracy becomes crucial. The ratio where you'd become indifferent depends on the scale and nature of the dataset, but using this formula helps find the tipping point for each group.
