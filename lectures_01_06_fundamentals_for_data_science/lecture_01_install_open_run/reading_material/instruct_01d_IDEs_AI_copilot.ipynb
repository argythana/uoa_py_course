# Lecture 1, instructions, part D, Integrated Development Environments (IDEs = more than "Editors") and coding co-pilots.

## 1) Install a modern IDE (VS Code or PyCharm)

### VS Code
1. Download VS Code from the official site: [https://code.visualstudio.com/](https://code.visualstudio.com/)
2. Install it using the default settings for your operating system.
3. Open VS Code and install these extensions:
   - Python (Microsoft)
   - Jupyter (Microsoft)
   - GitHub Copilot
   - GitHub Copilot Chat
4. Open your course folder in VS Code (`File -> Open Folder`).
5. Select the course Python environment/interpreter from the Python extension (Command Palette -> `Python: Select Interpreter`).

### PyCharm
[Integrated Development Environment: PyCharm IDE](https://www.jetbrains.com/pycharm/)
1. Install PyCharm Professional (or Community, if Professional license is not available).
2. If available, activate the Professional student license using your academic email.
3. Create a new project or open an existing course project.
4. Configure the Python interpreter for the project.
5. Explore [scientific tools](https://www.jetbrains.com/pycharm/features/scientific_tools.html).
6. Verify the selected Python interpreter version (bottom-right corner in PyCharm).

## 2) Integrate an AI coding co-pilot in your IDE

1. Sign up for the [GitHub Student Developer Pack](https://education.github.com/pack).
2. Activate your student GitHub Copilot access from [GitHub Copilot](https://github.com/features/copilot).
3. Sign in to GitHub inside your IDE.
4. Install and enable the GitHub Copilot extension/plugin:
   - VS Code: `GitHub Copilot` + `GitHub Copilot Chat` extensions
   - PyCharm: [GitHub Copilot plugin](https://plugins.jetbrains.com/plugin/17718-github-copilot)
5. Test it by creating a new Python file and asking Copilot to explain a simple function.

## How to use AI assistants during this course

Use AI as a learning assistant, not as an auto-pilot.

Try discussing the lecture notes with Copilot. Ask it to:
- explain lecture concepts in detail,
- identify mistakes, errors, or typos,
- improve clarity and structure,
- provide extra beginner and advanced examples.

During the whole course, use Copilot to:
- explain what is presented in the lecture notes,
- answer questions about lecture material,
- solve practice exercises,
- provide alternative solutions,
- compare different solution approaches,
- suggest more advanced exercises for practice.

**IMPORTANT**  

Using an IDE and an integrated AI co-pilot is **MANDATORY** for this course.
Please make sure your setup is ready before the second lecture.

### Please be aware:
1. AI hallucinations are real.
2. AI assistants are often not fully up to date (for example, deprecated APIs such as old pandas patterns).
3. You still need version control and careful review for AI-generated code.
4. Copy-paste is useful only when you understand what you copy.
5. AI-generated "slop" is real. Prioritize correctness, clarity, and testing.
